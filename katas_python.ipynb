```python
# 1.Escribe una función que reciba una cadena de texto como parámetro y devuelva un diccionario con las frecuencias de cada letra en la cadena. Los espacios no deben ser considerados.
def contador_frecuencia_letras(cadena):
    # Creamos un diccionario vacío
    frecuencias = {}
    # Pasamos el texto a minúsculas para que 'A' y 'a' se cuenten juntas
    cadena_limpia = cadena.lower()
    # Recorremos cada carácter de la cadena
    for letra in cadena_limpia:
        # Ponemos la condición para ignorar los espacios
        if letra != " ":
            # Si la letra ya está en el diccionario, le sumamos 1 a su contador
            if letra in frecuencias:
                frecuencias[letra] += 1
            # Si es la primera vez que vemos la letra, la añadimos con valor 1
            else:
                frecuencias[letra] = 1
                
    return frecuencias

# Ejecutamos la función para ver el resultado en Jupyter
texto = "Probando las katas"
print(contador_frecuencia_letras(texto))
```

```python
# 36. Función procesar_texto con subfunciones auxiliares y *args.

# Funciones auxiliares

def contar_palabras(texto):
    """Cuenta las palabras del texto."""
    
    palabras = texto.split()
    conteo = {}

    for palabra in palabras:
        # Quitamos signos y pasamos a minúsculas
        palabra = palabra.lower().strip(".,!?")
        conteo[palabra] = conteo.get(palabra, 0) + 1

    return conteo


def reemplazar_palabras(texto, palabra_original, palabra_nueva):
    """Reemplaza una palabra por otra."""
    
    return texto.replace(palabra_original, palabra_nueva)


def eliminar_palabra(texto, palabra_eliminar):
    """Elimina una palabra del texto."""

    palabras = texto.split()

    # Quitamos la palabra indicada
    palabras = [p for p in palabras if p != palabra_eliminar]

    return " ".join(palabras)


# Función principal

def procesar_texto(texto, opcion, *args):

    if opcion == "contar":
        return contar_palabras(texto)

    elif opcion == "reemplazar":

        palabra_original = args[0]
        palabra_nueva = args[1]

        return reemplazar_palabras(texto, palabra_original, palabra_nueva)

    elif opcion == "eliminar":

        palabra_eliminar = args[0]

        return eliminar_palabra(texto, palabra_eliminar)

    else:
        return "Opción no válida."


# Pruebas

texto_prueba = "el perro corre rapido por el parque el perro esta feliz"

print("Contar palabras")
print(procesar_texto(texto_prueba, "contar"))

print("\nReemplazar palabra")
print(procesar_texto(texto_prueba, "reemplazar", "perro", "gato"))

print("\nEliminar palabra")
print(procesar_texto(texto_prueba, "eliminar", "el"))
```

```python
# 39.Escribe una función que tome dos parámetros: figura (una cadena que puede ser "rectangulo", "circulo" o "triangulo") y datos (una tupla con los datos necesarios para calcular el área de la figura).

import math

# funcion para las distintas figuras y el calculo del area
def area(figura,datos):

    if figura == "rectangulo":
        base = datos[0]
        altura = datos[1]
        return base * altura
    elif figura == "triangulo":
        base = datos[0]
        altura = datos[1]
        return (base * altura) / 2
        
    elif figura == "circulo":
        radio = datos[0]  
        return math.pi * (radio ** 2)
        
    else:
        return "Figura no válida"
    
#  Pruba con un rectángulo, triangulo (base 6, altura 4) y circulo radio 6
figura1 = "rectangulo"
datos1 = (6, 4)
figura2 = "triangulo"
datos2 = (6, 4)
figura3 = "circulo"
datos3 = (6, )

print(f"El area del {figura1} es {area(figura1, datos1)}")
print(f"El area del {figura2} es {area(figura2, datos2)}")
print(f"El area del {figura3} es {area(figura3, datos3):.2f}")
```

```python
# 40.Escribe un programa en Python que utilice condicionales para determinar el monto final de una compra en una tienda en línea, después de aplicar un descuento.

# Pedimos el precio del artículo
precio = float(input("Introduce el precio del artículo: "))

# Preguntamos si tiene cupón
tiene_cupon = input("¿Tienes cupón de descuento? (si/no): ").lower()

descuento = 0

if tiene_cupon == "si" or tiene_cupon == "sí":

    # Pedimos el valor del cupón
    cupon = float(input("Introduce el valor del cupón: "))

    if cupon > 0:
        descuento = cupon
    else:
        print("Cupón no válido")

elif tiene_cupon == "no":
    descuento = 0

else:
    print("Opción no válida")

# Calculamos el precio final
precio_final = precio - descuento

# Evitamos precios negativos
if precio_final < 0:
    precio_final = 0

print(f"El precio final es: {precio_final:.2f} €")
```
